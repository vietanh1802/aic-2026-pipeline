```python
# 2. Cấu hình đường dẫn và lấy danh sách 10 ảnh đầu tiên
from pathlib import Path

IMG_DIR = Path('/content/drive/MyDrive/aic26/Extracted/K01')
IMG_EXTS = {'.jpg', '.jpeg', '.png', '.bmp', '.webp'}
N = 10

assert IMG_DIR.is_dir(), f'Khong tim thay thu muc: {IMG_DIR}'

# Sap xep theo ten (dung natural sort de 2 < 10) va loc dung file anh
import re
def natural_key(p):
    return [int(s) if s.isdigit() else s.lower() for s in re.split(r'(\d+)', p.name)]

image_paths = sorted(
    [p for p in IMG_DIR.iterdir() if p.suffix.lower() in IMG_EXTS],
    key=natural_key,
)[:N]

print(f'Tim thay {len(image_paths)} anh (hien thi toi da {N}):')
for p in image_paths:
    print(' -', p.name)
```

```python
# 3. Doc va hien thi 10 anh dau tien
import math
import matplotlib.pyplot as plt
from PIL import Image

cols = 5
rows = math.ceil(len(image_paths) / cols)
fig, axes = plt.subplots(rows, cols, figsize=(4 * cols, 4 * rows))
axes = axes.ravel() if len(image_paths) > 1 else [axes]

for ax, path in zip(axes, image_paths):
    img = Image.open(path).convert('RGB')
    ax.imshow(img)
    ax.set_title(f'{path.name}\n{img.size[0]}x{img.size[1]}', fontsize=9)
    ax.axis('off')

# An cac o thua (neu it hon rows*cols anh)
for ax in axes[len(image_paths):]:
    ax.axis('off')

plt.tight_layout()
plt.show()
```
